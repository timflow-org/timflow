# Test bessel performance

```python
from timflow.bessel.besselnumba import (
    besselld_gauss_ho_d1d2,
    besselld_gauss_ho_qxqy_d1d2,
    besselld_int_ho,
    besselld_int_ho_qxqy,
    bessells_gauss_ho_d1d2,
    bessells_gauss_ho_qxqy_d1d2,
    bessells_int_ho,
    bessells_int_ho_qxqy,
)


def timeit(func, N=1, *args, **kwargs):
    import time

    times = []
    start = time.time()
    for _ in range(N):
        result = func(*args, **kwargs)
        end = time.time()
        times.append(end - start)
    mean = sum(times) / len(times)
    std = (sum((x - mean) ** 2 for x in times) / len(times)) ** 0.5
    print(
        f"{mean:.2e} s ± {std:.2e} s per loop (mean ± std. dev.of 1 runs, {N} loops each)"
    )
    return result
```
